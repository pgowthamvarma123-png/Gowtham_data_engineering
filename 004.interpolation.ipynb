# Interpolation
process of embedding variables or expressions inside a string

```python
empname = "gowtham varma penmetsa"
```

```python
age = 29

print(f'employe name {empname} and age {age}')
```

```python
a = 20
b = 30
print(f" sum of a and b {a+b}")
```

```python
i = 1234.676868
print(i)
print(f'{i:.2f}')
```

```python
print(f'${i:.3f}')
```

```python
salary = 20000
print(f"{salary:.0%}")
```

```python
print("employee name : %s and age : %s" %(empname,age))
```
