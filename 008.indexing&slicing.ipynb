# Indexing and slicing

```python
a = "Gowtham varma"
print(a[4],a[5],end="")
print(a[0])
print(a[-1])
print(a[-5])
print(a[6])
```

```python
dob = "1996-10-22"
year = dob[0:4]
print(f'my birth year {year}')
print(len(dob))
```

```python
name = input('enter you name :')
len_dob = len(name)
print(f'my name : {name} \nname length : {len_dob}')
```

```python
k = 'misunderstanding'
print(k[13])
print(k[10])
print(k[-8])
print(k[-13])
print(k[3:5:1])
print(k[3:5])
print(k[5:])
print(k[5])
print(k[10:1:-1])
print(k[-11:-2])
print(k[-1::-1])
```

```python
a = "madam"

a[::] == a[::-1]
```
