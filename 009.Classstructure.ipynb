```python
class student():
    
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_info(self):
        print(f'student name :{self.name} student age :{self.age}')     
        
```

```python
student1 = student('gowtham',30)
```

```python
student1.display_info()
```

```python
class student_details():
    def __init__(self,num1,num2):
        self.num1 = num1
        self.num2 = num2
    def mult(self):
        total = self.num1* self.num2
        return total
    def sum_of_num(self):
        total_sum = self.num1 + self.num2
        return total_sum

        
```

```python
student1 = student_details(10,20)
```

```python
student1.__init__(10,20)
```

```python
print(student1.mult())
print(student1.sum_of_num())
```
