```python
# two table
number = int(input("enter number : "))
for i in range(1,11):
    print(f'{number} * {i} = {number*i}')
```

```python
# cummalative sum of numbers
num = int(input("enter number:"))
sum = 0
for i in range(1, num + 1):
    sum += i
    print(f"sum of numbers from 1 to {i} is: {sum}")
print(f"cumulative sum of numbers from 1 to {num} is: {sum}")
```

```python
# factorial of a number
num = int(input("enter a number: "))
fact = 1
for i in range(1,num+1):
     fact *= i
     print(f'factorial of {i} is : {fact}')
print(f'factorial total of {num} is : {fact}')

   
```

# While Loop 

```python
i = 1
while i <= 10 :
    print(i,end=' ')
    i += 1
```

```python
i = 5
while i >= 1:
    print('hello world',end=' ')
    i -= 1
```

```python
count = 1
total = 0
while count <= 10 :
    total += count 
    count += 1
    print(f'sum of numbers from 1 to {count} is : {total}',end=' ')
print(f'\ntotal sum of numbers from 1 to 10 is ;{total}')
```

```python
i = 0
while i < 10:
    i += 1
    if i <= 5:
        print(i,end='_')
        print(f'loop run {i}')
    else :
        break
print(f'break the loop at {i}')
```

```python
# continue statement will skip current iteration 

i = 0
while i < 10:
    i += 1
    if i == 5:
        print(f'current iteration skipped at {i}')
        continue
    print(i)
print(f'loop completed at {i}')
```
