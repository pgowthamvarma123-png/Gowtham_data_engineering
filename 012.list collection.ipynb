```python
#lists 

mynames = ["Alice", "Bob", "Charlie", "David"]
mynums = [1, 2, 3, 4, 5]
print(mynames)
print(mynums)
```

```python
# Range function will create a list of numbers from 10 to 50 with a step of 10

num = list(range(10,60,10))
print(num)
```

```python
# append() method adds an element to the end of the list
mynames.append('eve')
num.append(60)
```

```python
# insert() method adds an element at a specific index in the list
mynames.insert(1,'gowtham') 
num.insert(1,10)
```

```python
print(mynames,num)
```

```python
# Pop() method removes the last element from the list
mynames.pop()
num.pop()
```

```python
#remove() method removes the first occurrence of a specified value from the list
mynames.remove('gowtham')
num.remove(10)
```

```python
print(mynames,num)
```

```python
k = ['sunday','monday','tuesday']
y = []
for i in k:
    c = i[::-1] 
    result = ''
    for b in range(len(c)):
        if b % 3 == 0 :
            result += c[b].upper()
        else:
            result += c[b]
    y.append(result)
print(y)
```

```python
k = ['sunday','monday','tuesday']
y = []
for i in k:
    c = i[::-1] 
    result = ''
    for b in range(len(c)):
        if b % 3 == 0 :
            result += c[b].upper()
        else:
            continue
    y.append(result)
print(y)
```

```python
# Flatten a nested list using list comprehension
nested_list = [[1,2,3],[4,5,6],[7,8,9]]
flatten_list = [num for sublist in nested_list for num in sublist]
print(flatten_list)
```

```python
# length of the list

len_list = [len(sublist) for sublist in nested_list]
print(len_list)
```

```python
len_list = []
for sublist in nested_list:
    len_list.append(len(sublist))
print(len_list)
```

```python
len_dict = {i:len(sublist) for i,sublist in enumerate(nested_list)}
print(len_dict)
```

```python
# common elements in two lists
list1 = [1,2,3,4,5]
list2 = [3,4,5,6,7]

common_list = list(set(list1) & set(list2))
print(common_list)
```

```python
common_list2 = [value for value in list1 if value in list2]
print(common_list2)
```
