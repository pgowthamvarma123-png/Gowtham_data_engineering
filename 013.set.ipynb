# SET
set will maintain unique items
* immutable
* we cannot fetch index value

```python
# create Set

myset = {"apple", "banana", "cherry"}
myset = {"apple", "banana", "cherry", "apple", "banana"}
```

```python
print(myset)  # Output: {'banana', 'cherry', 'apple'}
```

```python
myset.add("orange")
```

```python
myset.union({"kiwi", "mango"})  
print(myset)  # Output: {'banana', 'cherry', 'apple', 'orange'}
```

```python
myset2 = myset.union({"kiwi", "mango"})  
print(myset2) 
```

```python
print(myset.difference(myset2))
```

```python
myset3 = myset2.difference(myset)
print(myset3)
```
