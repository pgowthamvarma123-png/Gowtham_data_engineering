```python
x = {}
n = int(input('number of elements :'))
for i in range(n):
    print('enter key : ')
    k = input()
    print('enter value :')
    v = input()
    x.update({k:v})
print('final dictionary list:',x)
```

```python
score = x.get('rohith',-1)
if score != -1:
    print(f'kohili score is :{score}')
else :
    print('player not found')
```

```python
colors = {}
n = int(input('number of elements :'))
for i in range(n):
    print('enter color code : ')
    k = input()
    print('enter color name:')
    v = input()
    colors.update({k:v})
print('final dictionary list:',colors)
for i in colors:
    print(i)
for j in colors:
    print(colors[j])
```

```python
for k,v in colors.items():
    print(f'key:{k} and value:{v}')
```

```python
# Comprehensions
numbers = [1,2,3,4,5,6]

squares = {
    num : num * num
    for num in numbers
}
print(squares)
```

```python
nums2 = {
    num * num for num in numbers
    
}
print(nums2)
```

```python
numbers = [1,2,3,4,5,6]

even_squares = {
    num : num * num
    for num in numbers
    if num % 2 == 0
}
print(even_squares)
```
