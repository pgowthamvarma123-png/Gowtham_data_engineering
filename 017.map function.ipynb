# Map Function
Create map and pass the variables from list one by one

```python
def calculation(x):
    return(x*2)
```

```python
num = list(range(10,50,10))
```

```python
mp = list(map(calculation,num))
print(mp)
```

```python
lambda x : x*2
print(x)
```

```python
mp = list(map(lambda x : x*2,num))
print(mp)
```
