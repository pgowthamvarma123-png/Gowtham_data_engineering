# Recursion 
means call the function repetatively inside the function itself

```python
def recursivenum (x):
    if x > 0 :
        print(x)
        return recursivenum(x-1)
```

```python
recursivenum(10)
```

```python
# cummalative sum
total = 0
def cummalativesum(x):
    if x == 0 or x == 1:
        return(x)
    else:
        total = 0
        for i in range(1,x+1):
            total += i
            print(total,end=' ')
    return(f'{total}')
```

```python
cummalativesum(20)
```

```python
def summnumrecursive(x):
    if x == 0 or x == 1 :
        return(x)
    print (x)
    return x + summnumrecursive(x-1)
```

```python
summnumrecursive(9)
```

```python
def factorialrecursive(x):
    if x == 0 or x == 1 :
        return(x)
    print (x)
    return x * factorialrecursive(x-1)
```

```python
factorialrecursive(10)
```

```python
def factorial(x):
    if x == 0 or x == 1:
        return(x)
    else:
        total = 1
        for i in range(1,x+1):
            total *= i
            print(total,end=' ')
    return(f'{total}')
```

```python
factorial(10)
```
