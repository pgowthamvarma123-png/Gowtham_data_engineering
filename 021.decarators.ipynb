# Decarators 
special function modifies the behaviour of another function without changing the code

```python
def greetings():
    return("hello world")
print(greetings())
```

```python
def uppercase_var(func):
    def inner():
        result = func()
        return result.upper()
    return inner
```

```python
@uppercase_var
def greetings():
    return("hello world")
print(greetings())
```

```python
def greet_decorator(func):
    def inner(name,num):
        print("Welcome!")
        print('hello gowtham')
        a = name
        a = a.upper()
        b = num
        print(a,'_',b)
        return a,b
    return inner
```

```python
@greet_decorator
def greet(name,num):
    return (name,num)
```
