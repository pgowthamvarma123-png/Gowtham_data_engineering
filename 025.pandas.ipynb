```python
# df.isnull().sum() will check the count of the none values

print(df.isnull().sum())
```

```python
#fillna function will fill the missing values with null for particular selected column

df['age'] = df['age'].fillna(df['age'].mean())
```

```python
df = df.dropna()
print(df)
```

```python
# groupby function like group clause in sql

grouped = df.groupby('city')
print(grouped)
```

```python
agg = df.groupby('city').agg({'age':['mean','min','max']})
print(agg)
```

```python
# Joining Data frames 
data2 = {'age2':[20,30,35,40],
         'Name' : ['charlie','bob','gowtham','varma']}
```

```python
df3 = pd.DataFrame(data2)
```

```python
mergerd = pd.merge(df,df3,on = "age")
print(mergerd)
```

```python
Joined_df = df.join(df3.set_index("Name"),on='Name')
print(Joined_df)
```

```python
pivoted = df.pivot(index='Name', columns='city', values='age')
print(pivoted)
```

```python
#melted function split the columns into row wise
meleted_df = df.melt(id_vars=['Name'],value_vars=['age','city'])
print(meleted_df)
```
