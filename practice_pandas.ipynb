```python
import pandas as pd
names = []
age = []
n = int(input("Enter the number of students: "))

for i in range(n):
    name = input("Enter the name of student {}: ".format(i+1))
    names.append(name)
    age_input = int(input("Enter the age of student {}: ".format(i+1)))
    age.append(age_input)

df = pd.DataFrame({'Name': names, 'Age': age})
print(df)
```

```python
f = open(r'C:\Users\gpenmetsa3\Desktop\gowtham\students.txt', 'w')
f.write(df.to_string(index=True))
f.close()
```
