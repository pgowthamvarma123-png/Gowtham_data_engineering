```python
def number_verify(a):
    if a > 0:
        print(f'input number {a} is positive')
    elif a < 0 :
        print(f'input number {a} is negative')
    else:
        print(f'input number {a} is zero')
```

```python
number_verify(0)
```

```python
def marks_verify(a):
    if a >= 90 :
        print('a grade')
    elif a >= 75:
        print('b grade')
    elif a >= 50 :
        print('c grade')
    else :
        print('fail')
```

```python
input_marks = int(input('enter marks : '))
marks_verify(input_marks)
```

```python
import pandas as pd 
n = int(input("Enter number of employees: "))

names = []
ages = []

for i in range(n):
    name = input(f"Enter name {i+1}: ")
    age = int(input(f"Enter age {i+1}: "))
    names.append(name)
    ages.append(age)

df = pd.DataFrame({
"Name": names,
"Age": ages
})

print("\nDataFrame:")
print(df)
```
